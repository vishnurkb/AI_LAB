```python
def dfs(state, path, visited):
    goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

    if state == goal:
        return path + [state]

    if len(path) >= 20:
        return None

    visited.add(state)

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = []
    if row > 0: moves.append(-3)
    if row < 2: moves.append(3)
    if col > 0: moves.append(-1)
    if col < 2: moves.append(1)

    for move in moves:
        new_blank = blank + move

        tiles = list(state)
        tiles[blank], tiles[new_blank] = tiles[new_blank], tiles[blank]
        new_state = tuple(tiles)

        if new_state not in visited:
            result = dfs(new_state, path + [state], visited.copy())
            if result:
                return result

    return None

input_data = input("Enter 9 numbers (use 0 for blank): ")
initial = tuple(map(int, input_data.split()))

if len(initial) != 9 or set(initial) != set(range(9)):
    print("Invalid input! Enter numbers from 0 to 8 exactly once.")
else:
    solution = dfs(initial, [], set())

    if solution:
        print("Solution found in", len(solution) - 1, "moves:")
        for state in solution:
            for i in range(0, 9, 3):
                print(*state[i:i+3])
            print()
    else:
        print("No solution found within 20 moves / Not possible")
```

```python
def dls(state, goal, depth, path):
    if state == goal:
        return path + [state]

    if depth == 0:
        return None

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = []
    if row > 0: moves.append(-3)
    if row < 2: moves.append(3)
    if col > 0: moves.append(-1)
    if col < 2: moves.append(1)

    for move in moves:
        new_zero = zero + move
        new_state = list(state)
        new_state[zero], new_state[new_zero] = new_state[new_zero], new_state[zero]
        new_state = tuple(new_state)

        if new_state not in path:
            result = dls(new_state, goal, depth - 1, path + [state])
            if result:
                return result

    return None


def ids(initial_state, goal):
    depth = 0
    while True:
        result = dls(initial_state, goal, depth, [])
        if result:
            return result
        depth += 1


input_data = input("Enter 9 numbers (use 0 for blank): ")
initial = tuple(map(int, input_data.split()))
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

if len(initial) != 9 or set(initial) != set(range(9)):
    print("Invalid input! Enter numbers from 0 to 8 exactly once.")
else:
    solution = ids(initial, goal)
    print("Solution found in", len(solution) - 1, "moves:")
    for state in solution:
        for i in range(0, 9, 3):
            print(*state[i:i+3])
        print()
```
