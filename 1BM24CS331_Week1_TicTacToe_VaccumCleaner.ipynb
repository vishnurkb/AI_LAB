```python
board = [" " for _ in range(9)]

def display_board():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()

def winner(player):
    positions = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, b, c in positions:
        if board[a] == board[b] == board[c] == player:
            return True
    return False

def is_draw():
    return " " not in board

def minimax(is_maximizing, depth):
    if winner("O"):
        return 10 - depth

    if winner("X"):
        return depth - 10

    if is_draw():
        return 0

    if is_maximizing:
        best_score = -float("inf")

        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(False, depth + 1)
                board[i] = " "
                best_score = max(best_score, score)

        return best_score

    best_score = float("inf")

    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            score = minimax(True, depth + 1)
            board[i] = " "
            best_score = min(best_score, score)

    return best_score

def best_move():
    best_score = -float("inf")
    move = -1

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(False, 1)
            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    return move, best_score

board = [
    "O", "O", "X",
    "X", " ", "O",
    " ", " ", "X"
]

display_board()

while not winner("O") and not winner("X") and not is_draw():
    move, score = best_move()
    board[move] = "O"

    display_board()

    if winner("O"):
        print("AI wins!")
        break

    if is_draw():
        print("Draw!")
        break

    print("AI score:", score)

    position = int(input("Enter your move (1-9): ")) - 1

    while position < 0 or position > 8 or board[position] != " ":
        print("Invalid move.")
        position = int(input("Enter your move (1-9): ")) - 1

    board[position] = "X"

    if winner("X"):
        display_board()
        print("You win!")
        break
```

```python
state = ["Dirty", "Dirty"]
position = 0

while True:

    if state[position] == "Dirty":
        print("Agent is in", "Left" if position == 0 else "Right", "room")
        print("Room is Dirty")
        print("Cleaning the room...")

        state[position] = "Clean"

        print("Room is now Clean")
        print()

    else:
        print("Agent is in", "Left" if position == 0 else "Right", "room")
        position = (position + 1) % len(state)

        print()
```

```python
state = ["Dirty", "Clean", "Dirty", "Dirty"]
room_length = len(state)
cost = 0
while True:

    for i in range(room_length):

        if state[i] == "Dirty":
            print("Agent moves to room", i + 1)
            print("Room is Dirty")
            print("Cleaning the room...")

            state[i] = "Clean"
            cost += 1
            print("Room is now Clean")
            print()

    if all(room == "Clean" for room in state):
        print("All rooms are Clean")
        print("Goal achieved!")
        print("Cost: ", cost)
        break
```
